Импорт библиотек

```python
import pandas as pd
import numpy as np
```

Загрузка данных

```python
df = pd.read_csv('train.csv')
df.head()
```

Основная информация о датасете.

```python
print("=== Форма датасета ===")
print(df.shape)

print("\n=== Типы данных ===")
print(df.dtypes)

print("\n=== Информация о датасете ===")
df.info()

print("\n=== Количество пропусков по столбцам ===")
print(df.isnull().sum())

print("\n=== Процент пропусков по столбцам ===")
print((df.isnull().sum() / len(df) * 100).round(2))

print("\n=== Описательная статистика (числовые) ===")
display(df.describe())

print("\n=== Описательная статистика (категориальные) ===")
display(df.describe(include='object'))
```

Процент выживаемости по классам

```python
survival_by_class = df.groupby('Pclass')['Survived'].mean() * 100
survival_by_class = survival_by_class.round(2).reset_index()
survival_by_class.columns = ['Pclass', 'Survival Rate (%)']
print("Процент выживаемости по классам пассажиров:")
```

Самые популярные имена на корабле

```python
def extract_first_name(full_name):
    after_comma = full_name.split(',')[1]
    after_dot = after_comma.split('.')[1]
    first_name = after_dot.strip().split()[0]
    return first_name

df['FirstName'] = df['Name'].apply(extract_first_name)

male_names = df[df['Sex'] == 'male']['FirstName']
female_names = df[df['Sex'] == 'female']['FirstName']

top_male = male_names.value_counts().idxmax()
top_female = female_names.value_counts().idxmax()

print(f"Самое популярное мужское имя на корабле: {top_male} "
      f"({male_names.value_counts().max()} чел.)")
print(f"Самое популярное женское имя на корабле: {top_female} "
      f"({female_names.value_counts().max()} чел.)")
```

Самые популярные имена на корабле

```python
for pclass in sorted(df['Pclass'].unique()):
    class_df = df[df['Pclass'] == pclass]

    top_m = class_df[class_df['Sex'] == 'male']['FirstName'].value_counts()
    top_f = class_df[class_df['Sex'] == 'female']['FirstName'].value_counts()

    top_m_name = top_m.idxmax() if not top_m.empty else 'N/A'
    top_f_name = top_f.idxmax() if not top_f.empty else 'N/A'

    print(f"  Класс {pclass}:")
    print(f"    Мужское имя:  {top_m_name} ({top_m.max() if not top_m.empty else 0} чел.)")
    print(f"    Женское имя:  {top_f_name} ({top_f.max() if not top_f.empty else 0} чел.)\n")
```

Пассажиры старше 44 лет

```python
older_than_44 = df[df['Age'] > 44]
print(f"Количество пассажиров старше 44 лет: {len(older_than_44)}")
```

Мужчины младше 44 лет

```python
young_males = df[(df['Age'] < 44) & (df['Sex'] == 'male')]
print(f"Количество мужчин младше 44 лет: {len(young_males)}")
```

Количество n-местных кабин

```python
cabin_counts = (
    df[df['Cabin'].notna()]
    .groupby('Cabin')
    .size()
    .reset_index(name='PassengersInCabin')
)

n_cabin_distribution = cabin_counts['PassengersInCabin'].value_counts().sort_index()
n_cabin_distribution.index.name = 'Вместимость (чел.)'
n_cabin_distribution.name = 'Количество кабин'

print("Количество n-местных кабин:")
display(n_cabin_distribution.to_frame())
```

Пассажиры без родственников

```python
# SibSp — братья/сёстры и супруги, Parch — родители и дети
no_relatives = df[(df['SibSp'] == 0) & (df['Parch'] == 0)]
print(f"Количество пассажиров без родственников на борту: {len(no_relatives)}")
```
